## More Advanced Functions

### Adding Documentation to functions using DocString

```python
# Example: DocString with function
def add(a, b):
    '''Adds two numbers together'''
    return a + b

add(1,2)
print(add.__doc__)
```

#### Also you can use `help` function

```python
help(add)
```

#### Also you can use `?` mark

```python
add?
```

```python
# Example: Doctring with class
class Person:
    '''A class to represent a Person'''
    pass

print(Person.__doc__)
```

```python
help(Person)
```

```python
Person?
```

### Dynamic Positional Arguments with `*args`

```python
# Example: Positional Arguments (*args)
def numbers(*args):
    print(args)

numbers()
numbers(10)
numbers(10, 20, 30)
```

```python
# Example: 
def add(*nums):
    result = 0
    for num in nums:
        result += num
    return result

add(10, 20, 30)
```

```python
L = [10, 20, 30, 40]

add(*L) # Un Packing
```

### Dynamic Keyword Arguments with `**kwargs`

```python
# Example: Keyword Arguments
def create_user_profile(**kwargs):
    print(kwargs)

create_user_profile(name='Ismail', age=23, city='Garowe')
```

```python
def create_user_profile(**kwargs):
    user_profile = {}
    for key, value in kwargs.items():
        user_profile[key] = value
    
    return user_profile

create_user_profile(name='Ismail', age=23, city='Garowe')
```

```python
D = {'name': 'Ismail', 'age': 23, 'city': 'Garowe'}

create_user_profile(**D) # Un Packing
```

#### Using both Positional Arguments and Keyword Arguments

```python
def print_Info(*args, **kwargs):
    print(args)
    print(kwargs)

print_Info(10, 20, name='Ismail', Age=23)
```

#### Using Decorator

```python
def debug(func):
    def wrapper(*args, **kwargs):
        print('Args:', args)
        print('Kwargs:', kwargs)

        return func(*args, **kwargs)
    
    return wrapper

@debug
def add(a, b):
    return a + b

add(10, 20)
```

```python
def Celsius_to_fahranheit(temp):
    '''Celcius to Fahranheit'''
    f = (temp * 9/5) + 32

    print(f'The Fahranhait is: {f}')

# Fahranhiet to Celcuis
def fahrenheit_to_Celsius(temp):
    '''fahrenheit to Clecius'''
    c = (temp - 32) * 5/9
    print(f'The Celsius is: {c}')

Celsius_to_fahranheit(25)
fahrenheit_to_Celsius(75)
```

```python
fahrenheit_to_Celsius(49)
Celsius_to_fahranheit(9.444444444444445)
```

```python
def my_decorator(func):
    def wrapper():
        print('Function-ka ka hor')
        func()
        print('Function-ka ka dib')

    return wrapper

@my_decorator
def salaan():
    print('Salaan!')

salaan()
```

```python
import time

def timer(func):
    def wrapper():
        start = time.time()
        func()
        end = time.time()
        print(f"Waqti: {end - start:.4f} ilbiriqsi")
    return wrapper

@timer
def shaqo():
    time.sleep(2)
    print("Shaqadii way dhammaatay")

shaqo()
```
