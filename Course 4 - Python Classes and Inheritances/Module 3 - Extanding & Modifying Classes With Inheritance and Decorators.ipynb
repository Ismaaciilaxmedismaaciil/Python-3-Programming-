## Inheritance

### Introduction: Class Inheritance
#### Inheritance is a fundemantal concept in object Oriented Programming that allows a class to inherit attribute and methods from another class.

Types of Inheritance: 
- Single Inheritance
- Multiple Inheritance

#### Type 1: Single Inheritance 

```python
# Inheritance

# Defining Parent Class
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    # Drive Method
    def drive(self):
        print(f'The Person will drive {self.enginetype} car.')


car1 = Car(4, 4, 'Petrol')
car1.drive()
```

```python
# Defining Inherit class or child class
class Tesla(Car):
    def __init__(self, doors, windows, enginetype, is_selfdriving):
        super().__init__(doors, windows, enginetype)
        self.is_selfdriving = is_selfdriving

    def selfdriving(self):
        print(f'Tesla supports selfdriving: {self.is_selfdriving}')

TeslaCyber = Tesla(4, 6, 'Electric', True)

TeslaCyber.selfdriving()
TeslaCyber.drive()
```

#### type 2: Multiple Inheritance

```python
# Example 2: Multiple Inheritance

# Base Class 1
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print('Subclasses must be implement this method')

# Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner

# Derived class / Inheriting both these classes
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f'{self.name} says woof!'

dog1 = Dog('Buddy', 'Krish')
print(dog1.speak())
print('Owner:', dog1.owner)
print('Name:', dog1.name)
```

```python
Current_year = 2026
# Parent Class
class Person:
    def __init__(self, name, year_born):
        self.name = name
        self.year_born = year_born
    
    def getAge(self):
        return Current_year - self.year_born

    def __str__(self):
        return f"{self.name} ({self.getAge()})"

# Child Class
class Student(Person):
    def __init__(self, name, year_born):
        Person.__init__(self, name, year_born)
        self.knowledge = 0

    
    def study(self):
        self.knowledge += 1
  
    
Ismail = Student('Ismail Ahmed', 2003)
print(Ismail)
Ismail.study()
print(Ismail.knowledge)
print(Ismail.getAge())
```

### Overriding Methods

```python
class Book():
    def __init__(self, title, author):
        self.title = title
        self.author = author

    def __str__(self):
        return f'"{self.title}" by {self.author}'


class PaperBook(Book):
    def __init__(self, title, author, numPages):
        Book.__init__(self, title, author)
        self.numPages = numPages

    def getNumPages(self):
        return f'{self.numPages} Pages'

class EBook(Book):
    def __init__(self, title, author, size):
        Book.__init__(self, title, author)
        self.size = size
    
    def getSize(self):
        return f'{self.size} MB'

class Library:
    def __init__(self):
        self.books = []

    def addBook(self, book):
        self.books.append(book)
    
    def getNumBooks(self):
        return f'{len(self.books)} books'

myPaperBook = PaperBook('The Odyssey', 'Ismail', 500)
print(myPaperBook)
print(myPaperBook.getNumPages())

myEBook = EBook('Get Rich', 'Steven', 2)
print(myEBook)
print(myEBook.getSize())


library1 = Library()
library1.addBook(myPaperBook)
library1.addBook(myEBook)
print(library1.getNumBooks())
```

```python
from random import randrange

# Here's the original Pet class
class Pet():
    boredom_decrement = 4
    hunger_decrement = 6
    boredom_threshold = 5
    hunger_threshold = 10
    sounds = ['Mrrp']
    def __init__(self, name = "Kitty"):
        self.name = name
        self.hunger = randrange(self.hunger_threshold)
        self.boredom = randrange(self.boredom_threshold)
        self.sounds = self.sounds[:]  # copy the class attribute, so that when we make changes to it, we won't affect the other Pets in the class

    def clock_tick(self):
        self.boredom += 1
        self.hunger += 1

    def mood(self):
        if self.hunger <= self.hunger_threshold and self.boredom <= self.boredom_threshold:
            return "happy"
        elif self.hunger > self.hunger_threshold:
            return "hungry"
        else:
            return "bored"

    def __str__(self):
        state = "     I'm " + self.name + ". "
        state += " I feel " + self.mood() + ". "
        # state += "Hunger %d Boredom %d Words %s" % (self.hunger, self.boredom, self.sounds)
        return state

    def hi(self):
        print(self.sounds[randrange(len(self.sounds))])
        self.reduce_boredom()

    def teach(self, word):
        self.sounds.append(word)
        self.reduce_boredom()

    def feed(self):
        self.reduce_hunger()

    def reduce_hunger(self):
        self.hunger = max(0, self.hunger - self.hunger_decrement)

    def reduce_boredom(self):
        self.boredom = max(0, self.boredom - self.boredom_decrement)
```

```python
class Cat(Pet):
    sounds = ['Meow']

    def mood(self):
        if self.hunger > self.hunger_threshold:
            return "hungry"
        if self.boredom <2:
            return "grumpy; leave me alone"
        elif self.boredom > self.boredom_threshold:
            return "bored"
        elif randrange(2) == 0:
            return "randomly annoyed"
        else:
            return "happy"

class Dog(Pet):
    sounds = ['Woof', 'Ruff']

    def mood(self):
        if (self.hunger > self.hunger_threshold) and (self.boredom > self.boredom_threshold):
            return "bored and hungry"
        else:
            return "happy"

c1 = Cat("Fluffy")
d1 = Dog("Astro")

c1.boredom = 1
print(c1.mood())
c1.boredom = 3
for i in range(10):
    print(c1.mood())
print(d1.mood())
```

### Invoking The Parents Classes Method

```python
class Person:
    def __init__(self, name):
        self.name = name
    
    def greet(self):
        return f'Hello, my name is {self.name}!'
    
class Student(Person):
    def __init__(self, name, grade):
        Person.__init__(self,name)
        self.grade = grade

    def greet(self):
        return Person.greet(self) + " I'm a student"
    
    def getGrade(self):
        return f'The grade is {self.grade}'
    
s = Student('Ismail', 12)
print(s.greet())

print(s.getGrade())
```

```python
# Using super keyword
class Person:
    def __init__(self, name):
        self.name = name
    
    def greet(self):
        return f'Hello, my name is {self.name}!'
    
class Student(Person):
    def __init__(self , name, grade):
        super().__init__(name)
        self.grade = grade
        
    def greet(self):
        return super().greet() + " I'm a student"
    
    def getGrade(self):
        return f'The grade is {self.grade}'
    
s = Student('Ismail', 12)
print(s.greet())
print(s.getGrade())
```
