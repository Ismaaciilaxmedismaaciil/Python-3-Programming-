## Inheritance

### Introduction: Class Inheritance
#### Inheritance is a fundemantal concept in object Oriented Programming that allows a class to inherit attribute and methods from another class.

Types of Inheritance: 
- Single Inheritance
- Multiple Inheritance

#### Type 1: Single Inheritance 

```python
# Inheritance

# Defining Parent Class
class Car:
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    # Drive Method
    def drive(self):
        print(f'The Person will drive {self.enginetype} car.')


car1 = Car(4, 4, 'Petrol')
car1.drive()
```

```python
# Defining Inherit class or child class
class Tesla(Car):
    def __init__(self, doors, windows, enginetype, is_selfdriving):
        super().__init__(doors, windows, enginetype)
        self.is_selfdriving = is_selfdriving

    def selfdriving(self):
        print(f'Tesla supports selfdriving: {self.is_selfdriving}')

TeslaCyber = Tesla(4, 6, 'Electric', True)

TeslaCyber.selfdriving()
TeslaCyber.drive()
```

#### type 2: Multiple Inheritance

```python
# Example 2: Multiple Inheritance

# Base Class 1
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print('Subclasses must be implement this method')

# Base class 2
class Pet:
    def __init__(self, owner):
        self.owner = owner

# Derived class / Inheriting both these classes
class Dog(Animal, Pet):
    def __init__(self, name, owner):
        Animal.__init__(self, name)
        Pet.__init__(self, owner)

    def speak(self):
        return f'{self.name} says woof!'

dog1 = Dog('Buddy', 'Krish')
print(dog1.speak())
print('Owner:', dog1.owner)
print('Name:', dog1.name)
```

```python
CURRENT_YEAR = 2026

# Parent Class
class Person:
    def __init__(self, name, year_born):
        self.name = name
        self.year_born = year_born

    def getAge(self):
        return (CURRENT_YEAR - self.year_born)
    



# Child Class
class Student(Person):
    def __init__(self, name, year_born, faculty, student_Id):
        Person.__init__(self, name, year_born)
        self.student_Id = student_Id
        self.faculty = faculty
    
    def Class_Study(self):
        self.faculty = 'Informatics'
        return f'{self.faculty}'
    
    def __str__(self):
        return f'Student name is {self.name}, his age is {self.getAge()}, and his faculty is {self.faculty}, and is Id is {self.student_Id}'

student1 = Student('Ismail', 2003, 'Informatics ','IT-1150')
print(student1)
```
