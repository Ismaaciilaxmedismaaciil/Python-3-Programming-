## Constructing Classes

### Class
#### Is a blue print for creating objects.
#### It has seperate attributes and seperate methods.

```python
# User Defined Classes
# Creating class names Point
class Point():
    
    # Method
    def getX(self):
        return self.x

# Instance Method
# Creating two instane methods
point1 = Point()
point2 = Point()

# Instance Variable
# Creating two instance variables
point1.x = 5
point2.x = 10

# Priting
print(point1.getX())
print(point2.getX())
```

```python
class Dog:
    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

dog1 = Dog('Buddy', 5)
print(dog1)
print(dog1.name)
print(dog1.age)

dog2 = Dog('Lucy', 3)
print(dog2)
print(dog2.name)
print(dog2.age)

print(dog1 == dog2)
```

```python
# Define Class with Methods
class Dog:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def bark(self):
        return f'{self.name} says woof!'

dog1 = Dog('Buddy', 5)
print(dog1.name)
print(dog1.age)
print(dog1.bark())
```

```python
# Modeling a Bank Account
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    # Deposit
    def deposit(self, amount):
        self.balance += amount
        print(f'{amount} is deposited. New balance is {self.balance}')
    
    # Withdrow
    def withdrow(self, amount):
        if amount > self.balance:
            print('Insufficant funds!') 
        else:
            self.balance -= amount
            print(f'{amount} is withdrown. New balance is {self.balance}')
    
    def get_balance(self):
        print(f'Your balance is {self.balance}')

# Creating an Account
Ismail = BankAccount('Ismail')

# Deposited new amount
Ismail.deposit(1000)

# Withdrown a new large amount
Ismail.withdrow(2000)

# With drown new amount
Ismail.withdrow(200)

# See the balance
Ismail.get_balance()
```
