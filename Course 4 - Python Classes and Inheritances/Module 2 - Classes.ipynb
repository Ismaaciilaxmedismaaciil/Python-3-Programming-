#### Check Your Understanding

Create a class called `Animal` that accepts two numbers as inputs and assigns them respectively to two instance variables: `arms` and `legs`. Create an instance method called `limbs` that, when called, returns the total number of limbs the animal has. To the variable name `spider`, assign an instance of `Animal` that has 4 arms and 4 legs. Call the limbs method on the `spider` instance and save the result to the variable name `spidlimbs`

```python
class Animal():
    def __init__(self, arms, legs):
        self.arms = arms
        self.legs = legs

    def limbs(self):
        return self.arms + self.legs
    
spider = Animal(4, 4)
spidlimbs = spider.limbs()
print(spidlimbs)
```

```python
# Modeling a Bank Account
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    # Deposit
    def deposit(self, amount):
        self.balance += amount
        print(f'{amount} is deposited. New balance is {self.balance}')
    
    # Withdrow
    def withdrow(self, amount):
        if amount > self.balance:
            print('Insufficant funds!') 
        else:
            self.balance -= amount
            print(f'{amount} is withdrown. New balance is {self.balance}')
    
    def get_balance(self):
        print(f'Your balance is {self.balance}')

# Creating an Account
Ismail = BankAccount('Ismail')

# Deposited new amount
Ismail.deposit(1000)

# Withdrown a new large amount
Ismail.withdrow(2000)

# With drown new amount
Ismail.withdrow(200)

# See the balance
Ismail.get_balance()
```

### Private and Public Instance Variables

| Public Variable | Private Variable |
|-----------------|------------------|
| `self.name` | `self.__name` |
| Meel kasta waa laga geli karaa | Si toos ah dibadda loogama geli karo |
| Waa la akhrisan karaa lana beddeli karaa (`obj.name`) | Si toos ah looma akhrisan karo ama looma beddeli karo (`obj.__name` wuxuu keenaa `AttributeError`) |
| Looma isticmaalo name mangling | Python waxay isticmaashaa **name mangling** oo u beddesha `_ClassName__variable` |
| Waxaa loo isticmaalaa xogta caadiga ah | Waxaa loo isticmaalaa xogta la doonayo in aan si toos ah loo taaban ama loo beddelin |
| Tusaale: `person.name` | Tusaale: `person._Person__name` (name mangling) |

#### Public Example

```python
class Person():
    def __init__(self, name, age):
        self.name = name
        self.age = age

    # Given name / FirstName
    def givenName(self):
        return self.name.split()[0]
    
    # Family name / LastName
    def familyName(self):
        return self.name.split()[-1]
    
    def isChild(self):
        return self.age < 18

person1 = Person('Ismail Ahmed', 23)
print(person1.givenName())
print(person1.familyName())
print(person1.isChild())

# Accessing Instance Variables
print(person1.name)
```

#### Private Example

```python
class Person:
    def __init__(self, name, age):
        self.__name = name
        self.__age = age

    # Given name / First Name
    def givenName(self):
        return self.__name.split()[0]

    # Family name / Last Name
    def familyName(self):
        return self.__name.split()[-1]

    # Is Child or Not
    def isChild(self):
        return self.__age < 18


person1 = Person("Ismail Ahmed", 23)

print(person1.givenName())      # Ismail
print(person1.familyName())     # Ahmed
print(person1.isChild())        # False

# Name mangling
print(person1._Person__name)    # Ismail Ahmed
```
